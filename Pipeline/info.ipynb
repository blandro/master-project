## Information on the pipeline

For the pipeline to be run, one first needs to install BLAST for the dedicated computer setup.
1. Go to [this](https://ftp.ncbi.nlm.nih.gov/blast/executables/blast+/LATEST/) page for installation.
    * In my case, at the time of installation, ncbi-blast-2.16.0+-win64.exe was used
2. Run the installer, install BLAST+ in a directory like
    * C:\Program Files\NCBI\blast-2.16.0+\bin
3. Add BLAST to System Path:
    * Environment Variables > System Variables > Path > Edit > New > C:\Program Files\NCBI\blast-2.16.0+\bin

Following, the desired proteome must be obtained. Either from your own files and data, or online.\
E.g. one can download a file from [UniProt](https://ftp.uniprot.org/pub/databases/uniprot/current_release/knowledgebase/).

Make sure if it is a zipped fasta, fasta.gz, unzip it, then place it in a desired folder.

The reference database from this pipeline is based on the composition of [TCDB](https://www.tcdb.org/), and primarily contains transporters obtained from bacteria and archaea, meaning it performs sub-optimal for eukaryotic samples. It also constitutes of reactions obtained from Rhea, mapping back to TCDB. This way, it manages to identify more reactions for transporters.
