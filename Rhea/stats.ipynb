# Statistics of Rhea

```python
import matplotlib.pyplot as plt
```

```python
categories = [
    "EC 1",
    "EC 2",
    "EC 3",
    "EC 4",
    "EC 5",
    "EC 6",
    "EC 7",
    "Unclassified"
]

values = [2565, 2403, 983, 837, 344, 274, 88, 9930]
```

```python
plt.figure(figsize=(12, 6))
plt.xticks(rotation=45, ha="right", fontsize=20)
plt.yticks(fontsize=20)
plt.xlabel("EC class", fontsize=20, labelpad=-25)
plt.ylabel("Number of reactions", fontsize=20)
plt.ylim(top=11000)
plt.grid(axis="y", linestyle="--", alpha=0.7)
blue40white = (0.6,0.6,1.0)
bars = plt.bar(categories, values, color=blue40white, edgecolor="black")
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 100, str(yval), ha="center", va="bottom", fontsize=20)


plt.tight_layout()
plt.savefig('ec_distribution_rhea.pdf')
plt.show()
```
