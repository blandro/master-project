# Rhea and TCDB in the final DB
This notebook quickly gets an overview of how much is provided from Rhea vs TCDB.

```python
import pandas as pd
import ast
```

```python
df = pd.read_csv("tcdb_rhea_either_reactions.tsv", sep="\t", usecols=[0,1,5,6])
df.drop_duplicates(inplace=True)
```

```python
def unique_rx(column):
    all_rx = []
    for cell in column:
        if pd.notna(cell):
            try:
                reactions = ast.literal_eval(cell)
                if isinstance(reactions, list):
                    all_rx.extend(reactions)
            except (ValueError, SyntaxError):
                pass
    return set(all_rx)

unique_tcdb_rx = unique_rx(df["TCDB:Reaction:CHEBI"])
unique_rhea_rx = unique_rx(df["Rhea:Reaction:CHEBI"])
```
