# Rhea and TCDB in the final DB
This notebook wquickly gets an overview of how much is provided from Rhea vs TCDB. Keep in mind that this is done before the subunital reactions have been shared across a complex. Thus it may be a tad imprecice.

```python
import pandas as pd
import numpy as np
import ast
```

```python
df = pd.read_csv("tcdb_rhea_either_reactions.tsv", sep="\t", usecols=[0,1,5,6])
df.drop_duplicates(inplace=True)
```

```python
all_rhea_reactions = []
for cell in df["Rhea:Reaction:CHEBI"]:
    if pd.notna(cell):
        try:
            reactions = ast.literal_eval(cell)
            if isinstance(reactions, list):
                all_rhea_reactions.extend(reactions)
        except (ValueError, SyntaxError):
            pass

unique_rhea_reactions = set(all_rhea_reactions)
len(unique_rhea_reactions)
```

```python
all_tcdb_reactions = []
for cell in df["TCDB:Reaction:CHEBI"]:
    if pd.notna(cell):
        try:
            reactions = ast.literal_eval(cell)
            if isinstance(reactions, list):
                all_tcdb_reactions.extend(reactions)
        except (ValueError, SyntaxError):
            pass

unique_tcdb_reactions = set(all_tcdb_reactions)
len(unique_tcdb_reactions)
unique_tcdb_reactions
```
