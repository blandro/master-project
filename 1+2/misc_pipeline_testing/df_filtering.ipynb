# DF Filtering and Reduction

This notebook aims to effectively adjust the output after the completed BLASTp. The columns that are to be kept, are the TCID, Query ID, Mechanism from TCDB, and R:Reaction, E-value and Identity.

```python
import pandas as pd
```

```python
cols_to_load = [1, 2, 9, 11, 14, 15, 23]
proteins = pd.read_csv("identified_proteins.tsv", sep="\t", usecols=cols_to_load, dtype=str)
proteins.drop_duplicates(inplace=True)
proteins.rename(columns={"Reaction":"TCDB:Reaction", "R:Equation":"Rhea:Reaction"}, inplace=True)
new_order = ["UID", "TCID", "QueryID", "TCDB:Reaction", "Rhea:Reaction", "Evalue", "Identity"]
proteins = proteins[new_order]
missing_reactions = proteins[proteins["TCDB:Reaction"].isna() & proteins["Rhea:Reaction"].isna()]
```
