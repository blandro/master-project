# CSV modification
This notebook's purpose it to modify the file obtained from query2 in UniProt_Rhea.ipynb. It is cleaner to do this here than in Approach 1/main1.ipynb.

```python
import pandas as pd

def extract_id(value):
    if pd.notna(value) and isinstance(value, str):
        return value.strip().split("/")[-1] 
    return ""

def process_csv(input_file, output_file):
    df = pd.read_csv(input_file)
    df = df.map(extract_id)
    df.columns = ["RID", "UID", "RDIR", "RSID"]
    df.to_csv(output_file, sep="\t", index=False)

process_csv("query2.csv", "uid_rhea_refseq.tsv")
```
